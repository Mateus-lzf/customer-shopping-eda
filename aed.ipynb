**Limpeza e Qualidade dos Dados**

Antes de criar os gráficos, é necessário garantir a qualidade dos dados. Foram identificados três problemas:

**Duplicatas:** 50 linhas completamente idênticas, incluindo o `Customer ID`. 
Como o mesmo cliente não pode ter duas transações exatamente iguais, 
foram removidas. Shape: (5050, 17) → (5000, 17)

**Valores nulos:** quatro colunas com dados faltantes:
- `Purchase Amount`, `Previous Purchases`, `Review Rating` → preenchidos com a **mediana**, por ser mais robusta que a média em dados com outliers
- `Size` → preenchido com a **moda**, por ser uma variável categórica

**Inconsistência:** a coluna `Season` continha "Fall" e "Autumn" representando 
a mesma estação. Os valores foram unificados para "Autumn".

```python
# Removendo duplicatas 
df = df.drop_duplicates().copy()
print(f"Shape após remover duplicatas: {df.shape}")
```

```python
# Preenchendo numéricas com mediana
for col in ['Purchase Amount (USD)', 'Previous Purchases', 'Review Rating']:
    mediana = df[col].median()
    df[col] = df[col].fillna(mediana)
    print(f"{col} → mediana: {mediana}")

# Preenchendo categórica com moda
moda_size = df['Size'].mode()[0]
df['Size'] = df['Size'].fillna(moda_size)
print(f"Size → moda: {moda_size}")

# Confirmação
print(f"\nNulos restantes: {df.isnull().sum().sum()}")
```

```python
# Padronizando 'Season'
df['Season'] = df['Season'].replace('Fall', 'Autumn')
```

**Criando visualizações**

**Perfil dos Clientes**

Quem são os clientes dessa loja? As análises abaixo exploram idade, gênero 
e localização para entender o público-alvo.

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Age', bins=20, kde=True, color='teal')
plt.title('Distribuição de Idade dos Clientes')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.show()
```

> A base de clientes é bem distribuída entre 18 e 70 anos, sem concentração 
em nenhuma faixa etária específica.

```python
plt.figure(figsize=(6, 6))
plt.pie(df['Gender'].value_counts().values, 
        labels=df['Gender'].value_counts().index,
        autopct='%1.1f%%', colors=['#5DADE2', '#EC7063' ,'#009E73'])
plt.title('Distribuição de Gênero dos Clientes')
plt.show()
```

> A maior parte dos clientes são do gênero masculino (aproximadamente 60%), enquanto 33% são do gênero feminino

```python
plt.figure(figsize=(12, 6))
top_locations = df['Location'].value_counts().head(10)
sns.barplot(x=top_locations.index, y=top_locations.values, hue=top_locations.index, palette='mako')
plt.title('Top 10 Localizações dos Clientes')
plt.xlabel('Localização')
plt.ylabel('Contagem')
plt.xticks(rotation=45)
plt.show()
```

> New York lidera com folga como principal mercado, com 40% mais clientes 
que Los Angeles. É possível identificar dois grupos distintos: grandes mercados 
(NY, LA, Houston, Phoenix, Chicago) e mercados secundários com volume 
significativamente menor.

**Análise de Vendas**

Quais categorias e produtos movimentam mais a loja?

```python
receita_categoria = df.groupby('Category')['Purchase Amount (USD)'].sum()
receita_categoria = receita_categoria.sort_values(ascending=True)
plt.figure(figsize=(10, 6))
plt.barh(receita_categoria.index, receita_categoria.values, color='midnightblue')
plt.gca()
plt.xlim(0, 300000)
plt.title('Total de Gastos por Categoria')
plt.xlabel('Total de Gastos (USD)')
plt.ylabel('Categoria')

for barra in plt.gca().patches:
    plt.text(barra.get_width() + 2000, 
             barra.get_y() + barra.get_height()/2,
             f'${barra.get_width():,.0f}', 
             va='center', fontsize=10)

plt.show()
```

> Roupas lidera a receita com $266k, seguido por Acessórios ($213k) e 
Eletrônicos ($148k). Calçados e Agasalhos juntos representam menos de 10% 
do faturamento total, sugerindo oportunidade de expansão ou redução de 
estoque nessas categorias.

```python
produtos_mais_vendidos = df['Item Purchased'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=produtos_mais_vendidos.index, y=produtos_mais_vendidos.values, palette='mako', hue=produtos_mais_vendidos.index, legend=False)
plt.title('Top 10 Produtos Mais Vendidos')
plt.xlabel('Produto')
plt.ylabel('Quantidade Vendida')
plt.show()
```

> Shirt e Shoes lideram as vendas com ~315 e ~305 unidades respectivamente, 
bem acima dos demais que ficam equilibrados em torno de 165-170 unidades. Esses dois itens 
são os carros-chefe da loja — a estratégia pode focar em garantir estoque 
desses produtos ou investir em marketing dos demais para equilibrar o portfólio.

**Comportamento de Compra**

Como, quando e quanto os clientes gastam?

```python
ticket_medio_por_metodo_pagamento = df.groupby('Payment Method')['Purchase Amount (USD)'].mean().sort_values(ascending=False)
plt.figure(figsize=(10,6))
ax = sns.barplot(x=ticket_medio_por_metodo_pagamento.index, y=ticket_medio_por_metodo_pagamento.values, palette='mako', hue=ticket_medio_por_metodo_pagamento.index, legend=False)
plt.title('Ticket Médio por Método de Pagamento')
plt.xlabel('Método de Pagamento')
plt.ylabel('Ticket Médio (USD)')

for barra in ax.patches:
    ax.text(barra.get_x() + barra.get_width() / 2,
            barra.get_height() + 3,
            f'${barra.get_height():,.0f}',
            ha='center', fontsize=10)

plt.show()
```

> Clientes que pagam com cartão de débito têm ticket médio de $178, 
quase 3x maior que os usuários de Venmo ($59). Isso pode indicar que 
compras de maior valor tendem a ser pagas no débito — o método de pagamento 
pode ser um reflexo do tipo de produto comprado, não necessariamente do perfil 
do cliente.

```python
compras_por_estacao = df.groupby('Season')['Purchase Amount (USD)'].sum().reindex(['Spring', 'Summer', 'Autumn', 'Winter'])  # Garantindo a ordem correta
plt.figure(figsize=(10,6))
ax = sns.lineplot(x=compras_por_estacao.index, y=compras_por_estacao.values, marker='o', color='teal')
plt.fill_between(compras_por_estacao.index, compras_por_estacao.values, alpha=0.3)
plt.title('Total de Compras por Estação')
plt.xlabel('Estação')
plt.ylabel('Total de Compras (USD)')
plt.ylim(0, 220000)

for i, (estacao, valor) in enumerate(compras_por_estacao.items()):
    ax.text(i, valor + 5000, f'${valor:,.0f}', ha='center', fontsize=9)

plt.show()
```

> As vendas se mantêm relativamente estáveis ao longo das estações, com 
Outono liderando levemente ($191k) e Verão com o menor valor ($158k). 
A ausência de sazonalidade forte sugere que a loja não depende de datas 
sazonais para vender, o que indica uma base de clientes consistente ao longo do ano.
